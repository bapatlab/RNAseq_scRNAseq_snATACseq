```r
library(ArchR)
```

Specify fragment file names

```r
inputFiles <- c("fragment_files/fragments_lfd.tsv.gz", "fragment_files/fragments_hfd.tsv.gz")
names(inputFiles) <- c("LFD","HFD")
```

Import barcodes for cells that passed the cellranger QC filter (~ 3000 for each sample)

We do this because we know that the cellranger QC is "industry standard" and that the barcodes that pass the QC are the ones that we want to keep. It just saves us some further processing.

```r
# Read the filtered barcodes with full paths
hfd_filtered_barcodes <- readLines("/Users/scc/Google Drive/My Drive/Curio Science/BaparLab_Sequencing/snATACseq/HFD/outs/filtered_peak_bc_matrix/barcodes.tsv")
lfd_filtered_barcodes <- readLines("/Users/scc/Google Drive/My Drive/Curio Science/BaparLab_Sequencing/snATACseq/LFD/outs/filtered_peak_bc_matrix/barcodes.tsv")

# Check how many barcodes we have
length(hfd_filtered_barcodes)
length(lfd_filtered_barcodes)
```

```r
addArchRThreads(threads = 16)
addArchRGenome("mm10")
```

```r
# Create Arrow files with specified barcodes and very low thresholds (QC already done by only using the barcodes that passed the cellranger QC)
ArrowFiles <- createArrowFiles(
  inputFiles = c("fragments_hfd.tsv.gz", "fragments_lfd.tsv.gz"),
  sampleNames = c("HFD", "LFD"),
  validBarcodes = list(
    HFD = hfd_filtered_barcodes,
    LFD = lfd_filtered_barcodes
  ),
  minTSS = 1,
  minFrags = 100,
  addTileMat = TRUE,
  addGeneScoreMat = TRUE,
  force = TRUE
)
```

```r
print(ArrowFiles)
```
